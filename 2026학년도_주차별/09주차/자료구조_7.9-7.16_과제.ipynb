## 7.9 삽입 정렬의 빈칸


````python
def insertion_sort(A):
    n = len(A)
    for i in range(1, n):
        key = A[i]
        j = i - 1
        while j >= 0 and A[j] > key:
            A[j + 1] = A[j]
            j -= 1
        A[j + 1] = key
````

key보다 큰 항목을 한 칸씩 뒤로 옮긴 후 빈자리에 key를 넣는다.

## 7.11 삽입 정렬의 안정성


삽입할 값보다 큰 항목만 뒤로 옮기고 같은 값은 옮기지 않는다. 따라서 같은 키를 가진 항목들의 원래 상대적인 순서가 정렬 후에도 유지되므로 안정 정렬이다.


## 7.12 삽입 정렬의 최선·최악 입력


최선: 이미 오름차순으로 정렬된 경우이다. 각 항목을 앞의 항목과 비교한 뒤 바로 제자리에 두므로 시간 복잡도는 O(n)이다.

최악: 내림차순으로 정렬된 경우이다. 각 항목을 맨 앞까지 옮겨야 하므로 이동 횟수는 1 + 2 + ... + (n - 1)이고, 시간 복잡도는 O(n²)이다.

## 7.15 선형 조사법의 삭제 문제와 해결 방법


충돌로 뒤에 저장된 항목이 있을 때 앞 항목을 그냥 빈 칸으로 바꾸면, 이후 탐색이 그 빈 칸에서 멈춰 뒤의 항목을 찾지 못할 수 있다. 예를 들어 같은 해시 주소에서 시작하는 12가 1번, 23이 2번에 저장되어 있을 때 12를 단순 삭제하면 23을 탐색하다 1번의 빈 칸에서 실패한다.

삭제한 위치를 완전히 비우지 않고 삭제 표시(DELETED)로 남긴다. 탐색은 삭제 표시를 만나도 계속하고, 삽입할 때는 해당 칸을 재사용할 수 있다. 필요하면 테이블을 재구성하는 방법도 있다.


## 7.16 선형 조사법의 탐색 종료 조건


1. 찾는 키와 같은 항목을 만나면 탐색 성공으로 종료한다.

2. 한 번도 사용하지 않은 빈 칸을 만나면 해당 키가 없으므로 탐색 실패로 종료한다.

3. 모든 칸을 조사했거나 시작 위치로 한 바퀴 돌아왔는데도 키를 찾지 못하면 탐색 실패로 종료한다.

삭제 표시가 된 칸은 원래 빈 칸과 다르므로 탐색을 계속한다. 배열의 마지막 칸에 도착했을 때에는 0번 칸으로 돌아가 조사하며, 마지막 칸이라는 이유만으로 종료하지 않는다.
