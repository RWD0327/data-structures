### 문제-7

#### 두 개의 정수를 매개변수로 받아 최댓값과 최솟값을 함께 반환하는 함수를 작성하시오. 반환된 두 값을 출력하시오.

### 문제-8
[ 변수의 범위 ]
##### 다음 코드에서 오류가 발생하는 이유를 설명하고, global을 사용하여 정상적으로 실행되도록 수정하시오.


```python
# 원본 소스
count = 0

def increase():
    count = count + 1

increase()
print(count)
```

```python
# 함수 안에서 count에 대입하면 지역 변수로 인식되어, 대입 전 count를 참조하므로 오류가 발생한다.
# global을 사용하면 전역 변수 count를 수정할 수 있다.
count = 0

def increase():
    global count
    count = count + 1

increase()
print(count)
```

### 문제-9
[ 클래스와 생성자 ]
##### 다음 조건을 만족하는 Car 클래스를 작성하시오.

- color, speed를 인스턴스 변수로 가짐
- 생성자에서 두 값을 전달받음
- speedUp() 호출 시 속도 10 증가
- speedDown() 호출 시 속도 10 감소
- 자동차의 색상과 속도를 출력하는 show() 메서드 구현


```python
class Car:
    def __init__(self, color, speed):
        self.color = color
        self.speed = speed

    def speedUp(self):
        self.speed += 10

    def speedDown(self):
        self.speed -= 10

    def show(self):
        print("색상:", self.color)
        print("속도:", self.speed)

#---------------------------------

car = Car("빨강", 50)
car.speedUp()
car.speedDown()
car.speedDown()
car.show()
```

## 문제-10
[ 상속과 메서드 재정의 ]
##### 다음 조건을 만족하는 클래스를 작성하시오.

Animal 클래스를 정의하고 다음 멤버변수를 포함하시오.
 - name: 동물 이름
 - age: 동물 나이

Animal 클래스에 생성자 __init__()를 구현하시오.

Animal 클래스에 speak() 메서드를 구현하시오.
 -   "동물이 소리를 냅니다."를 출력

Dog 클래스가 Animal 클래스를 상속하도록 작성하시오.

Dog 클래스에 다음 멤버변수를 추가하시오.
 - color: 동물 색상
 - owner: 주인 이름

Dog 클래스에서 speak() 메서드를 재정의하여 "멍멍!"을 출력하시오.

Dog 객체를 생성하고 모든 멤버변수의 값을 출력하시오.

Dog 객체의 speak() 메서드를 호출하시오.


```python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def speak(self):
        print("동물이 소리를 냅니다.")

class Dog(Animal):
    def __init__(self, name, age, color, owner):
        super().__init__(name, age)
        self.color = color
        self.owner = owner

    def speak(self):
        print("멍멍!")

dog = Dog("초코", 3, "갈색", "Kim")
print("이름:", dog.name)
print("나이:", dog.age)
print("색상:", dog.color)
print("주인:", dog.owner)
dog.speak()
```
